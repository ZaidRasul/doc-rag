## Data Ingestion

```python
from langchain_core.documents import Document
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import PyMuPDFL oader, PyPDFLoader
```

```python
doc = Document(
    page_content="This is a test document for testing purposes at start of this notebook.",
    metadata={"source": "test_source", "author": "test_author", "pages":1 },
)
doc
```

```python
loader = TextLoader("../data/text-files/python-intro.txt", encoding="utf8")# it creates text loader object which can be 
#used to load text files into Document objects.
document = loader.load() # it loads the text file into a list of Document objects.
print(document)
```

```python
# load multiple text files from a directory

dir_loader = DirectoryLoader("../data/text-files/", glob="*.txt", loader_cls=TextLoader) # it creates directory loader object which can be used to load all the text files in the directory into Document objects.

dir_doc = dir_loader.load() # it loads all the text files in the directory into a list of Document objects.
dir_doc # returns a list of Document objects, each representing a text file in the directory.
```

```python
pdf_loader = PyMuPDFLoader("../data/pdf/ml-notes.pdf") # it creates pdf loader object
pdf_doc = pdf_loader.load()
pdf_doc
```
